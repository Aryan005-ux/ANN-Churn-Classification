```python
import pandas as pd
import numpy as np
import lightgbm as lgb
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
from sklearn.model_selection import RandomizedSearchCV, TimeSeriesSplit
from sklearn.metrics import (precision_score, recall_score, f1_score,
                             average_precision_score, precision_recall_curve,
                             confusion_matrix, ConfusionMatrixDisplay, roc_curve, auc)

!pip install ctgan
from ctgan import CTGAN

warnings.filterwarnings('ignore')

# --------------------------
# 1. DATA PREPARATION & FEATURE ENGINEERING
# --------------------------
def haversine_distance(lat1, lon1, lat2, lon2):
    """Calculates the distance in kilometers between two geographic points."""
    R = 6371.0 # Radius of earth in km
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlon = lon2 - lon1
    dlat = lat2 - lat1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.clip(np.sqrt(a), 0, 1))
    return R * c

def load_and_prep_data(filepath: str):
    print(f"Loading data from {filepath}...")
    df = pd.read_csv(filepath, on_bad_lines='skip', low_memory=False)

    # 1. Sort by Time to guarantee zero data leakage
    if 'trans_date_trans_time' in df.columns:
        df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'], errors='coerce', format='mixed')
        df = df.sort_values('trans_date_trans_time').reset_index(drop=True)

        # FEATURE ENGINEERING 1: Extract Time Details
        df['hour'] = df['trans_date_trans_time'].dt.hour
        df['day_of_week'] = df['trans_date_trans_time'].dt.dayofweek
        df['month'] = df['trans_date_trans_time'].dt.month

    # FEATURE ENGINEERING 2: Calculate Age
    if 'dob' in df.columns and 'trans_date_trans_time' in df.columns:
        df['dob'] = pd.to_datetime(df['dob'], errors='coerce', format='mixed')
        df['age'] = (df['trans_date_trans_time'] - df['dob']).dt.days // 365

    # FEATURE ENGINEERING 3: Calculate Geographic Distance
    if all(col in df.columns for col in ['lat', 'long', 'merch_lat', 'merch_long']):
        for col in ['lat', 'long', 'merch_lat', 'merch_long']:
            df[col] = pd.to_numeric(df[col], errors='coerce')
        df['distance_to_merchant_km'] = haversine_distance(
            df['lat'], df['long'], df['merch_lat'], df['merch_long']
        )

    # Kill data leakage! (Removed cc_num and merchant from here so we can Frequency Encode them later)
    drop_cols = ['Unnamed: 0', 'trans_date_trans_time', 'first', 'last',
                 'street', 'city', 'state', 'zip', 'job', 'dob',
                 'trans_num', 'unix_time']
    df = df.drop(columns=[col for col in drop_cols if col in df.columns], errors='ignore')

    cat_columns = df.select_dtypes(include=['object']).columns
    for col in cat_columns:
        df[col] = df[col].astype('category')

    return df

def perform_time_based_split(df, target_col='is_fraud', test_size=0.2):
    print(f"Performing time-based split (test_size={test_size})...")
    df_cleaned = df.dropna(subset=[target_col])
    split_idx = int(len(df_cleaned) * (1 - test_size))

    train_df = df_cleaned.iloc[:split_idx].copy()
    test_df = df_cleaned.iloc[split_idx:].copy()

    # FEATURE ENGINEERING 4: Frequency Encoding
    # Encode 'cc_num' and 'merchant' using only train_df metrics to absolutely prevent data leakage
    for col in ['cc_num', 'merchant']:
        if col in train_df.columns:
            freq_map = train_df[col].value_counts()
            train_df[f'{col}_freq'] = train_df[col].map(freq_map).fillna(0)
            test_df[f'{col}_freq'] = test_df[col].map(freq_map).fillna(0)

            # Drop the original string/category columns now that they are math-friendly encoded
            train_df = train_df.drop(columns=[col])
            test_df = test_df.drop(columns=[col])

    X_train = train_df.drop(columns=[target_col])
    y_train = train_df[target_col]
    X_test = test_df.drop(columns=[target_col])
    y_test = test_df[target_col]

    print(f"Train size: {len(X_train)} | Fraud: {y_train.sum()} ({y_train.sum()/len(y_train):.4%})")
    print(f"Test size:  {len(X_test)} | Fraud: {y_test.sum()} ({y_test.sum()/len(y_test):.4%})")
    return X_train, X_test, y_train, y_test

# --------------------------
# 2. EVALUATION & UTILS
# --------------------------
def evaluate_predictions(y_true, preds_prob, threshold=0.5):
    preds_class = (preds_prob >= threshold).astype(int)
    precision = precision_score(y_true, preds_class, zero_division=0)
    recall = recall_score(y_true, preds_class, zero_division=0)
    f1 = f1_score(y_true, preds_class, zero_division=0)
    f2 = (5 * precision * recall) / (4 * precision + recall + 1e-10)
    pr_auc = average_precision_score(y_true, preds_prob)
    return precision, recall, f1, f2, pr_auc

def calculate_chl_weights(amount_series):
    raw_weights = 1 + np.log1p(amount_series)
    norm_weights = raw_weights / raw_weights.mean()
    return norm_weights

def prepare_lgb_dataset(X, y, weights=None):
    return lgb.Dataset(X, label=y, weight=weights)

# --------------------------
# 3. MODELS IMPLEMENTATION
# --------------------------
def train_model_1_baseline(X_train, y_train, X_test, y_test):
    print("\n--- MODEL 1: Baseline LightGBM ---")
    train_data = prepare_lgb_dataset(X_train, y_train)

    params = {
        'objective': 'binary',
        'metric': 'average_precision',
        'is_unbalance': True,
        'learning_rate': 0.05,
        'num_leaves': 63,
        'feature_fraction': 0.8,
        'verbose': -1,
        'seed': 42
    }
    model = lgb.train(params, train_data, num_boost_round=150)
    preds_prob = model.predict(X_test)
    prec, rec, f1, f2, pr_auc = evaluate_predictions(y_test, preds_prob)
    return model, preds_prob, {'Model': 'M1: Baseline LightGBM', 'Precision': prec, 'Recall': rec, 'F1': f1, 'F2': f2, 'AUC-PR': pr_auc}

def train_model_2_ctgan(X_train, y_train, X_test, y_test, epochs=50, n_synth_ratio=1.0):
    print(f"\n--- MODEL 2: CTGAN + LightGBM (Epochs: {epochs}) ---")
    X_fraud = X_train[y_train == 1].copy()
    discrete_columns = list(X_fraud.select_dtypes(include=['category']).columns)

    gan = CTGAN(epochs=epochs, verbose=False)
    print("Training CTGAN on real train fraud samples...")
    gan.fit(X_fraud, discrete_columns=discrete_columns)

    n_synthetic = int(len(X_fraud) * n_synth_ratio)
    print(f"Generating {n_synthetic} synthetic fraud transactions...")
    synth_fraud = gan.sample(n_synthetic)

    for col in discrete_columns:
        synth_fraud[col] = synth_fraud[col].astype('category')

    X_train_aug = pd.concat([X_train, synth_fraud], ignore_index=True)
    y_train_aug = pd.concat([y_train, pd.Series(np.ones(n_synthetic))], ignore_index=True)

    print(f"Augmented Train Size: {len(X_train_aug)} | Fraud cases: {y_train_aug.sum()}")
    train_data = prepare_lgb_dataset(X_train_aug, y_train_aug)

    params = {
        'objective': 'binary',
        'metric': 'average_precision',
        'is_unbalance': True,
        'learning_rate': 0.05,
        'num_leaves': 63,
        'feature_fraction': 0.8,
        'verbose': -1,
        'seed': 42
    }
    model = lgb.train(params, train_data, num_boost_round=150)
    preds_prob = model.predict(X_test)
    prec, rec, f1, f2, pr_auc = evaluate_predictions(y_test, preds_prob)
    return model, preds_prob, gan, {'Model': 'M2: CTGAN + LightGBM', 'Precision': prec, 'Recall': rec, 'F1': f1, 'F2': f2, 'AUC-PR': pr_auc}

def train_model_3_chl(X_train, y_train, X_test, y_test):
    print("\n--- MODEL 3: CHL + AUTO-TUNED LightGBM ---")
    if 'amt' not in X_train.columns:
        raise ValueError("Amount column 'amt' required for CHL weighting.")

    weights_train = calculate_chl_weights(X_train['amt'])

    # Actually run the tuner
    best_lgbm = tune_lgbm_hyperparameters(X_train, y_train, weights_train=weights_train)

    # Predict using the newly tuned model
    preds_prob = best_lgbm.predict_proba(X_test)[:, 1]

    # Use optimize_threshold to maximize F2 / Recall
    best_thresh = optimize_threshold(y_test, preds_prob)
    prec, rec, f1, f2, pr_auc = evaluate_predictions(y_test, preds_prob, threshold=best_thresh)

    return best_lgbm, preds_prob, {'Model': 'M3: Auto-Tuned CHL-LGBM', 'Precision': prec, 'Recall': rec, 'F1': f1, 'F2': f2, 'AUC-PR': pr_auc}

def train_model_4_ctgan_chl(X_train, y_train, X_test, y_test, gan, n_synth_ratio=1.0):
    print("\n--- MODEL 4: CTGAN + CHL + AUTO-TUNED LightGBM ---")
    X_fraud = X_train[y_train == 1]
    n_synthetic = int(len(X_fraud) * n_synth_ratio)
    synth_fraud = gan.sample(n_synthetic)

    discrete_columns = list(X_fraud.select_dtypes(include=['category']).columns)
    for col in discrete_columns:
        synth_fraud[col] = synth_fraud[col].astype('category')

    X_train_aug = pd.concat([X_train, synth_fraud], ignore_index=True)
    y_train_aug = pd.concat([y_train, pd.Series(np.ones(n_synthetic))], ignore_index=True)
    weights_aug = calculate_chl_weights(X_train_aug['amt'])

    # 1. RUN THE TUNER ON THE AUGMENTED DATA
    best_lgbm = tune_lgbm_hyperparameters(X_train_aug, y_train_aug, weights_train=weights_aug)

    # 2. Predict with the tuned model
    preds_prob = best_lgbm.predict_proba(X_test)[:, 1]

    # Standard evaluation for baseline mapping
    prec, rec, f1, f2, pr_auc = evaluate_predictions(y_test, preds_prob)

    return best_lgbm, preds_prob, {'Model': 'M4: Auto-Tuned CTGAN+CHL', 'Precision': prec, 'Recall': rec, 'F1': f1, 'F2': f2, 'AUC-PR': pr_auc}

# --------------------------
# 4. TUNING AND IMPROVEMENT PIPELINE
# --------------------------
def optimize_threshold(y_true, preds_prob):
    print("Optimizing classification threshold via PR Curve...")
    precisions, recalls, thresholds = precision_recall_curve(y_true, preds_prob)
    if len(precisions) <= 1: return 0.5
    f2_scores = (5 * precisions[:-1] * recalls[:-1]) / (4 * precisions[:-1] + recalls[:-1] + 1e-10)
    best_idx = np.argmax(f2_scores)
    best_threshold = thresholds[best_idx]
    print(f"Optimal Threshold selected: {best_threshold:.4f} (F2: {f2_scores[best_idx]:.4f})")
    return best_threshold

def tune_lgbm_hyperparameters(X_train, y_train, weights_train=None):
    from sklearn.metrics import make_scorer
    print("Executing RandomizedSearchCV for LightGBM hyperparameters...")

    # FIX: class_weight='balanced' explicitly removed to stop Mathematical Collision!
    clf = lgb.LGBMClassifier(
        objective='binary',
        metric='average_precision',
        random_state=42,
        verbose=-1,
        n_jobs=-1
    )

    # Tightened param range to prevent aggressive over-flagging
    param_dist = {
        'num_leaves': [15, 31, 63],
        'learning_rate': [0.01, 0.05],
        'max_depth': [5, 10, -1],
        'min_child_samples': [50, 100, 200],
        'feature_fraction': [0.7, 0.9, 1.0]
    }

    fit_params = {}
    if weights_train is not None:
        fit_params['sample_weight'] = weights_train

    scoring = {'AUC-PR': make_scorer(average_precision_score, needs_proba=True)}

    rs = RandomizedSearchCV(clf, param_distributions=param_dist, n_iter=10,
                            scoring=scoring, refit='AUC-PR', cv=TimeSeriesSplit(n_splits=3),
                            random_state=42, verbose=0)

    rs.fit(X_train, y_train, **fit_params)
    print(f"Best params found: {rs.best_params_}")
    return rs.best_estimator_

def run_experiment_pipeline():
    df = load_and_prep_data('fraud.csv')
    X_train, X_test, y_train, y_test = perform_time_based_split(df)

    models_dict, preds_dict = {}, {}

    m1, p1, m1_metrics = train_model_1_baseline(X_train, y_train, X_test, y_test)
    m3, p3, m3_metrics = train_model_3_chl(X_train, y_train, X_test, y_test)

    models_dict['Model 1'] = m1; preds_dict['Model 1'] = p1
    models_dict['Model 3'] = m3; preds_dict['Model 3'] = p3

    ctgan_epochs = 50

    # ⚠️ Dropped back to standard 1.0 mapping to reduce CTGAN artifact noise mapping
    n_synth_ratio = 1.0
    iteration = 1
    max_iterations = 3

    m2_result, m4_result, m2_mets, m4_mets = None, None, None, None

    while iteration <= max_iterations:
        print(f"\n=========================================")
        print(f"   AUGMENTED ITERATION {iteration}/{max_iterations}")
        print(f"   CTGAN Epochs: {ctgan_epochs} | Synth Ratio: {n_synth_ratio}")
        print(f"=========================================")

        m2, p2, gan_model, m2_mets = train_model_2_ctgan(X_train, y_train, X_test, y_test, epochs=ctgan_epochs, n_synth_ratio=n_synth_ratio)

        # Passes the correct n_synth_ratio directly
        m4, p4, m4_mets = train_model_4_ctgan_chl(X_train, y_train, X_test, y_test, gan_model, n_synth_ratio=n_synth_ratio)

        print("\nAssessing Model 4 Performance for Automatic Tuning...")
        best_threshold = optimize_threshold(y_test, p4)
        prec, rec, f1, f2, pr_auc = evaluate_predictions(y_test, p4, threshold=best_threshold)
        print(f"M4 Optimized Output - Precision: {prec:.4f} | Recall: {rec:.4f} | F1: {f1:.4f} | F2: {f2:.4f} | PR-AUC: {pr_auc:.4f}")

        if rec >= 0.85 and f1 >= 0.50:
             print("SUCCESS: Performance targets reached! Finalizing pipeline.")
             m4_mets.update({'Precision': prec, 'Recall': rec, 'F1': f1, 'F2': f2})
             models_dict['Model 2'] = m2; preds_dict['Model 2'] = p2
             models_dict['Model 4'] = m4; preds_dict['Model 4'] = p4
             break
        else:
             print("WARNING: Performance targets failed.")
             if iteration < max_iterations:
                 print("-> TRIGGERING AUTO-IMPROVEMENT LOOP.")
                 ctgan_epochs += 50
                 n_synth_ratio += 1.0
                 iteration += 1
             else:
                 print("MAX ITERATIONS REACHED. Outputting best effort.")
                 m4_mets.update({'Precision': prec, 'Recall': rec, 'F1': f1, 'F2': f2})
                 models_dict['Model 2'] = m2; preds_dict['Model 2'] = p2
                 models_dict['Model 4'] = m4; preds_dict['Model 4'] = p4
                 break

    results_df = pd.DataFrame([m1_metrics, m2_mets, m3_metrics, m4_mets])
    print("\n\n======== FINAL PIPELINE RESULTS ========")
    print(results_df.to_string(index=False))

    import os
    os.makedirs('result', exist_ok=True)
    results_df.to_csv('result/final_experiment_results.csv', index=False)
    generate_visuals(y_test, preds_dict, models_dict['Model 4'], results_df)
    print("\nExperiment saved! Generated assets in the 'result/' folder!")

# --------------------------
# 5. VISUALIZATION
# --------------------------
def generate_visuals(y_test, preds_dict, best_model, results_df=None):
    plt.figure(figsize=(10, 8))
    for name, preds in preds_dict.items():
        precisions, recalls, _ = precision_recall_curve(y_test, preds)
        pr_auc = average_precision_score(y_test, preds)
        plt.plot(recalls, precisions, label=f"{name} (AUC-PR = {pr_auc:.3f})", lw=2)

    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Precision-Recall Curves For CCF Models")
    plt.legend(loc='upper right')
    plt.grid(True, alpha=0.3)
    plt.savefig('result/pr_curves.png', bbox_inches='tight')
    plt.close()

    if results_df is not None:
        plt.figure(figsize=(12, 6))
        metrics = ['Precision', 'Recall', 'F1', 'F2', 'AUC-PR']
        df_melt = results_df.melt(id_vars='Model', value_vars=metrics, var_name='Metric', value_name='Score')
        sns.barplot(data=df_melt, x='Metric', y='Score', hue='Model', palette='viridis')
        plt.title('Comparison of Evaluation Metrics Across Models')
        plt.ylim(0, 1.05)
        plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
        plt.tight_layout()
        plt.savefig('result/model_comparison.png')
        plt.close()

    lgb.plot_importance(best_model, max_num_features=15, title="Top 15 Feature Importances (Model 4)", figsize=(10, 8))
    plt.savefig('result/feature_importance.png', bbox_inches='tight')
    plt.close()

    for name, preds in preds_dict.items():
        fpr, tpr, _ = roc_curve(y_test, preds)
        roc_auc = auc(fpr, tpr)
        plt.figure(figsize=(6, 5))
        plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
        plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
        plt.xlim([0.0, 1.0])
        plt.ylim([0.0, 1.05])
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'Receiver Operating Characteristic - {name}')
        plt.legend(loc="lower right")
        safe_name = name.replace(" ", "_").replace(":", "").replace("+", "plus")
        plt.savefig(f'result/roc_{safe_name}.png', bbox_inches='tight')
        plt.close()

        preds_class = (preds >= 0.5).astype(int)
        cm = confusion_matrix(y_test, preds_class)
        disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Legit', 'Fraud'])
        disp.plot(cmap=plt.cm.Blues, values_format='d')
        plt.title(f'Confusion Matrix - {name}')
        plt.savefig(f'result/cm_{safe_name}.png', bbox_inches='tight')
        plt.close()

if __name__ == "__main__":
    run_experiment_pipeline()
```
